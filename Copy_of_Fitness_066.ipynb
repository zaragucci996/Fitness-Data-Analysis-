```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('06_fitness_data.csv')
df.isnull().sum()
df.duplicated().sum()
df[df['age']<0]
df[df['age']>80]
df[df['exercise_hours']<0]
df[df['exercise_hours']>24]
df[df['fitness_score']>100]
df[df['fitness_score']<0]
df[df['steps_per_day']<0]
df[df['steps_per_day']>50000]
df[df['water_liters']<0]
df[df['water_liters']>10]
sns.boxplot(y=df['age'])
plt.show()
sns.boxplot(y=df['exercise_hours'])
plt.show()
sns.boxplot(y=df['fitness_score'])
plt.show()
sns.boxplot(y=df['steps_per_day'])
plt.show()
sns.boxplot(y=df['water_liters'])
plt.show()
```
