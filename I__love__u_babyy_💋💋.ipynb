```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from IPython.display import Video, display

# Settings
FPS = 24
SECONDS = 8
FRAMES = FPS * SECONDS
OUTPUT = "love_heart_animation.mp4"

# Heart shape
t = np.linspace(0, 2*np.pi, 700)

x = 16 * np.sin(t)**3
y = (13*np.cos(t)
     - 5*np.cos(2*t)
     - 2*np.cos(3*t)
     - np.cos(4*t))

# Scale
x = x / 18.5
y = y / 18.5

# Text positions
idx = np.linspace(0, len(t)-1, 120, dtype=int)
px = x[idx]
py = y[idx]

# Figure
fig, ax = plt.subplots(figsize=(10, 7))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")

ax.set_xlim(-1.45, 1.45)
ax.set_ylim(-1.25, 1.25)
ax.set_aspect("equal")
ax.axis("off")

# Title
ax.text(
    0, 1.12,
    "Love Heart Animation",
    ha="center",
    fontsize=25,
    color="#f5c9ee",
    fontweight="bold"
)

# Create "I love you" texts
texts = []

for xx, yy in zip(px, py):

    txt = ax.text(
        xx, yy,
        "I love you",
        ha="center",
        va="center",
        fontsize=8,
        color="#f2a9d9",
        alpha=0.2
    )

    texts.append(txt)

# Center text
center = ax.text(
    0, 0,
    "I love you",
    ha="center",
    va="center",
    fontsize=20,
    color="#ffd5ef",
    alpha=0,
    fontweight="bold"
)

# Animation function
def update(frame):

    phase = frame / FRAMES * 2 * np.pi

    # Heart beating effect
    pulse = 1 + 0.055 * np.sin(phase)

    for i, (txt, xx, yy) in enumerate(zip(texts, px, py)):

        # Move heart
        txt.set_position(
            (xx * pulse, yy * pulse)
        )

        # Glow/shimmer
        wave = (
            np.sin(phase * 2 + i * 0.42) + 1
        ) / 2

        txt.set_alpha(
            0.2 + 0.75 * wave
        )

        txt.set_fontsize(
            7.5 + 2 * wave
        )

    # Center glow
    center.set_alpha(
        0.15 + 0.35 * max(0, np.sin(phase))
    )

    return texts + [center]


# Create animation
anim = FuncAnimation(
    fig,
    update,
    frames=FRAMES,
    interval=1000/FPS,
    blit=True
)

# Save as MP4
writer = FFMpegWriter(
    fps=FPS,
    bitrate=2500
)

anim.save(
    OUTPUT,
    writer=writer,
    dpi=120
)

plt.close(fig)

print("❤️ Video successfully created!")
print(OUTPUT)

# Show video
display(
    Video(
        OUTPUT,
        embed=True
    )
)
```
